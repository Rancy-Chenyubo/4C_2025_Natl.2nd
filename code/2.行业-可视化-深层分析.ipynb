平均创新费用

```python
import matplotlib.pyplot as plt
from matplotlib import font_manager
# 设置支持中文的字体
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']  
plt.rcParams['axes.unicode_minus'] = False 
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 读取数据
df = pd.read_csv(r'/home/mw/input/data_total74037403/行业角度数据集/全国企业创新调查-主表.csv')

years = [2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
data = []
for year in years:
    df_year = df[df['年份'] == year]
    sum_by_type = df_year.groupby('企业规模')['创新费用支出合计(亿元)'].sum()
    data.append(sum_by_type)

# 对齐索引，确保所有企业规模都能显示
types = sorted(set().union(*[d.index for d in data]))
data = [d.reindex(types, fill_value=0) for d in data]

# 计算占比
ratios = [d/d.sum() for d in data]

color_sets = [
    ['#E3EAF6', "#F9DCDC", "#DEF3E9", "#EBE0F1"],  # 2016
    ['#D0DFF7', "#FACFCF", "#D6F5E6", "#E9D6F4"],  # 2017
    ["#BFCEED", "#FFB3B3", "#C5EAD3", "#D2BAE7"],  # 2018
    ["#A4BCE8", "#FFA0A0", "#C7E6D2", "#D9A7F0"],  # 2019
    ["#96B0E7", "#FC8E8E", "#A9E5BE", "#B47FDF"],  # 2020
    ["#598AEC", "#F17474", "#9DDCB5", "#AE44E3"],  # 2021
    ["#4F65E0", "#EF6056", "#65BA9A", "#8E55D3"],  # 2022
    ["#1650CE", "#F33D3D", "#35A070", "#7E08E6"],  # 2023
]

fig, ax = plt.subplots(figsize=(12, 12))

# 调小环宽
width = 0.07
# 缩小中间空白圆的范围
inner_radius = 0.10
radii = [inner_radius + i*width for i in range(len(years))]

for i in range(len(years)-1, -1, -1): 
    wedges, texts, autotexts = ax.pie(
        ratios[i], radius=radii[i]+width/2, labels=types if i == len(years)-1 else None,
        autopct='%1.1f%%', colors=color_sets[i],
        wedgeprops=dict(width=width, edgecolor='w'), startangle=90,
        pctdistance=0.85 if i == len(years)-1 else 0.75
    )
    # 微调“微型企业”标签位置
    if i == len(years)-1:
        for j, text in enumerate(texts):
            if text.get_text() == "微型企业":
                x, y = text.get_position()
                text.set_position((x, y-0.04)) 

plt.title('2016-2023年各规模企业创新费用占比（多层环形图）')
plt.axis('equal')
plt.show()
```

创新阻碍

```python
import pandas as pd
import matplotlib.pyplot as plt

# 读取数据
df = pd.read_csv('/home/mw/input/data_total74037403/行业角度数据集/创新阻碍.csv')

scales = ['大型企业', '中型企业', '小型企业', '微型企业']
titles = ['大型企业', '中型企业', '小型企业', '微型企业']

fig, axes = plt.subplots(2, 2, figsize=(16, 10))
axes = axes.flatten()

for i, scale in enumerate(scales):
    row = df[(df['年份'] == 2023) & (df['企业规模'] == scale)].iloc[0]
    data = row[2:]
    # 颜色列表：大于8为浅绿色，否则为蓝色
    bar_colors = ['#93CBA4' if v > 8 else '#4F77CE' for v in data.values]
    bars = axes[i].barh(data.index, data.values, color=bar_colors)
    axes[i].set_title(f'2023年{scale}创新阻碍', fontsize=14)
    axes[i].set_xlim(0, max(data.values)*1.15)
    for y, v in enumerate(data.values):
        axes[i].text(v + 0.5, y, f'{v:.1f}', va='center', fontsize=10)
    axes[i].invert_yaxis()
    axes[i].set_xlabel('占比（%）')
    if i % 2 == 0:
        axes[i].set_ylabel('阻碍类型')
    else:
        axes[i].set_ylabel('')

plt.tight_layout()
plt.show()
```
