```python
import pandas as pd
import jieba
from wordcloud import WordCloud  
import matplotlib.pyplot as plt  
from collections import Counter  
import numpy as np
from PIL import Image

# 解决中文显示问题  
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# 1. 加载停用词表  
def load_stopwords(path):
    with open(path, encoding='utf-8') as f:
        return set(line.strip() for line in f if line.strip())

all_stopwords = load_stopwords('/home/mw/input/data_total74037403/个人角度数据集/停用词表.txt')

# 文件读取（智能编码检测）
def read_weibo_data(file_path): 
    encodings = ['utf-8', 'gb18030', 'gbk', 'ISO-8859-1']
    for encoding in encodings:
        try:
            return pd.read_csv(file_path, encoding=encoding)
        except UnicodeDecodeError:
            continue
    raise ValueError("无法自动检测文件编码，请手动指定")

data = read_weibo_data('/home/mw/input/data_total74037403/个人角度数据集/weibo_data.csv')

def process_text(text):  
    # 非字母数字替换为空格
    text = ''.join([char if char.isalnum() else ' ' for char in text])  
    words = jieba.lcut(text)  
    return [w for w in words if len(w) > 1 and w not in all_stopwords]

# 2. 读取评论并分词统计  
comment_list = data['内容'].dropna().astype(str).tolist()  
all_words = []
for comment in comment_list:
    all_words.extend(process_text(comment))

word_counts = Counter(all_words)
top50 = word_counts.most_common(50)

print("【微博评论高频词TOP50】")
for i, (w, c) in enumerate(top50, 1):
    print(f"{i:2d}. {w:<8} {c:>6}次")

# 3. 加载遮罩图（AI.JPG），并生成词云  
mask_img = np.array(Image.open('/home/mw/input/data_total74037403/个人角度数据集/ai1.jpg'))


wc = WordCloud(
    width=1600,
    height=1200,
    background_color='white',
    max_words=200,
    mask=mask_img,
    colormap='viridis',
    scale=2,
).generate_from_frequencies(word_counts)

plt.figure(figsize=(20, 12))
plt.imshow(wc, interpolation='bilinear')
plt.title("微博生成式AI讨论热点分析", fontsize=30, pad=20)
plt.axis('off')

# 保存
wc.to_file('weibo_ai_wordcloud.png')
plt.savefig('wordcloud.jpg', dpi=300, bbox_inches='tight')
plt.show()
```
