```python
import pandas as pd
import seaborn as sns

import matplotlib.pyplot as plt

# 设置全局字体为微软雅黑
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# 导入数据
file_path = r'/home/mw/input/data_total74037403/国家角度数据集/fig_1.2.3.csv'
data = pd.read_csv(file_path)

# 数据预处理
data['Year'] = data['Year'].astype(int)
data['Number of AI patent filings (in thousands)'] = data['Number of AI patent filings (in thousands)'].astype(float)

# 可视化分析
plt.figure(figsize=(16, 8))
sns.set_theme(style="whitegrid")

# 绘制折线图
ax = sns.lineplot(
    data=data,
    x='Year',
    y='Number of AI patent filings (in thousands)',
    hue='Geographic area',
    style='Application status',
    markers=True,
    dashes=False,
    palette='Set2'
)

# 添加数据标签
for line in ax.lines:
    for x, y in zip(line.get_xdata(), line.get_ydata()):
        if not pd.isna(y):  # 过滤掉 NaN 值
            ax.text(x, y, f'{y:.1f}', fontsize=8, ha='center', va='bottom')

# 设置标题和标签
plt.title('AI Patent Filings by Year, Application Status, and Geographic Area', fontsize=16)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Number of AI Patent Filings (in thousands)', fontsize=12)
plt.legend(title='Legend', fontsize=10, title_fontsize=12)
plt.tight_layout()

# 显示图表
plt.show()
```

```python
import pandas as pd
import seaborn as sns

import matplotlib.pyplot as plt

# 设置全局字体为微软雅黑
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# 数据导入
file_path = r'/home/mw/input/data_total74037403/国家角度数据集/fig_1.2.4.csv'
data = pd.read_csv(file_path)

# 数据预处理
data_pivot = data.pivot(index='Year', columns='Region', values='Granted AI patents (% of world total)')

# 绘制折线图
plt.figure(figsize=(12, 8))
for region in data_pivot.columns:
    plt.plot(data_pivot.index, data_pivot[region], marker='o', label=region)

    # 添加数据标签
    for x, y in zip(data_pivot.index, data_pivot[region]):
        if not pd.isna(y):
            plt.text(x, y, f'{y:.4f}', fontsize=8, ha='center', va='bottom')

# 图表美化
plt.title('各地区 AI 专利占全球总量的百分比（2010-2022）', fontsize=16)
plt.xlabel('年份', fontsize=12)
plt.ylabel('专利占比 (%)', fontsize=12)
plt.legend(title='地区', fontsize=10, title_fontsize=12)
plt.grid(alpha=0.3)
plt.tight_layout()

# 显示图表
plt.show()

# 绘制复杂美观的表格
plt.figure(figsize=(14, 6))
sns.heatmap(data_pivot, annot=True, fmt='.4f', cmap='YlGnBu', cbar_kws={'label': '专利占比 (%)'})
plt.title('各地区 AI 专利占全球总量的百分比热力图（2010-2022）', fontsize=16)
plt.xlabel('地区', fontsize=12)
plt.ylabel('年份', fontsize=12)
plt.tight_layout()

# 显示热力图
plt.show()
```

```python
import pandas as pd
import seaborn as sns

import matplotlib.pyplot as plt

# 设置全局字体为微软雅黑
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# 导入数据
file_path = r'/home/mw/input/data_total74037403/国家角度数据集/fig_1.3.3.csv'
data = pd.read_csv(file_path)

# 数据预处理
pivot_data = data.pivot(index='Year', columns='Label', values='Number of notable machine learning models')

# 绘制折线图
plt.figure(figsize=(12, 8))
for country in pivot_data.columns:
    plt.plot(pivot_data.index, pivot_data[country], marker='o', label=country)

# 添加数据标签
for country in pivot_data.columns:
    for x, y in zip(pivot_data.index, pivot_data[country]):
        plt.text(x, y, str(y), fontsize=8, ha='center', va='bottom')

# 图表美化
plt.title('Number of Notable Machine Learning Models by Region (2003-2023)', fontsize=16)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Number of Models', fontsize=12)
plt.legend(title='Region', fontsize=10)
plt.grid(visible=True, linestyle='--', alpha=0.6)

# 显示图表
plt.tight_layout()
plt.show()
plt.figure(figsize=(12, 6))
sns.heatmap(pivot_data, annot=True, fmt='d', cmap='YlGnBu', linewidths=0.5, linecolor='gray', cbar_kws={'label': 'Number of Models'})
plt.title('Heatmap of Notable Machine Learning Models by Region (2003-2023)', fontsize=16)
plt.xlabel('Region', fontsize=12)
plt.ylabel('Year', fontsize=12)

# 显示热力图
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

data = pd.read_csv('/home/mw/input/data_total74037403/国家角度数据集/fig_1.3.3.csv')
pivot_data = data.pivot(index='Year', columns='Label', values='Number of notable machine learning models')

plt.figure(figsize=(12, 8))

years = pivot_data.index
bar_width = 0.35

main_countries = list(pivot_data.columns[:2])
colors = ['#ffb6c1', "#ade6b6"]  # 浅粉色、浅蓝色

# 英文转中文映射
country_map = {
    'United States': '美国',
    'China': '中国',
    'United Kingdom': '英国',
    'Canada': '加拿大',
    'France': '法国',
    'Germany': '德国',
    'Japan': '日本',
    'South Korea': '韩国',
    'European Union and United Kingdom' : '欧盟和英国',
    'Other': '其他'
}

# 替换主国家为中文
main_countries_cn = [country_map.get(c, c) for c in main_countries]
other_countries = pivot_data.columns[2:]
other_countries_cn = [country_map.get(c, c) for c in other_countries]

for i, country in enumerate(main_countries):
    plt.bar(
        years + i*bar_width - bar_width/2,  # 偏移
        pivot_data[country],
        width=bar_width,
        label=main_countries_cn[i],
        color=colors[i],
        edgecolor='white'
    )

# 其余国家用绿色折线，标签为中文
for idx, country in enumerate(other_countries):
    plt.plot(years, pivot_data[country], marker='o', label=other_countries_cn[idx], color='lightblue')

# 添加数据标签（仅主柱形）
for i, country in enumerate(main_countries):
    for x, y in zip(years + i*bar_width - bar_width/2, pivot_data[country]):
        plt.text(x, y, str(y), fontsize=8, ha='center', va='bottom')

plt.title('各地区重要机器学习模型数量（2003-2023年）', fontsize=16)
plt.xlabel('年份', fontsize=12)
plt.ylabel('模型数量', fontsize=12)
plt.legend(title='地区', fontsize=10)
plt.grid(visible=True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from matplotlib.ticker import MaxNLocator
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

file_path = r'/home/mw/input/data_total74037403/国家角度数据集/fig_1.5.3.csv'
data = pd.read_csv(file_path)

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(data['Year'], data['Number of GitHub stars (in millions)'], color='skyblue', edgecolor='black')


for i, value in enumerate(data['Number of GitHub stars (in millions)']):
    ax.text(data['Year'][i], value + 0.1, f'{value:.2f}', ha='center', fontsize=10)

ax.set_title('GitHub Stars Growth Over Years', fontsize=16, fontweight='bold')
ax.set_xlabel('Year', fontsize=12)
ax.set_ylabel('Number of GitHub Stars (in millions)', fontsize=12)
ax.xaxis.set_major_locator(MaxNLocator(integer=True))
ax.grid(axis='y', linestyle='--', alpha=0.7)


plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.3)
plt.show()
```
