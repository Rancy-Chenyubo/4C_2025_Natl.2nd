```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

data = {
    "项目": ["合计", "考察访问", "国际会议", "合作研究", "培训", "展览会", "其他"],
    "2005": [19132, 6218, 6565, 2341, 1593, 577, 1838],
    "2010": [40572, 8316, 17549, 5575, 2635, 487, 6010],
    "2015": [72103, 11136, 34794, 12123, 2954, 472, 10624],
    "2020": [11850, 1342, 4963, 2239, 1018, 36, 2252],
    "2021": [13461, 1532, 6724, 1938, 1962, 46, 1259],
    "2022": [14188, 1476, 6021, 2298, 2476, 97, 1820],
    "2023": [76358, 12262, 40255, 12601, 3314, 485, 7441]
}

df = pd.DataFrame(data)
df.set_index("项目", inplace=True)

linestyles = ['-', '--', '-.', ':', (0, (3, 1, 1, 1)), (0, (5, 5))]
colors = plt.cm.tab10.colors

fig, ax = plt.subplots(figsize=(12, 6))



xlim = (0, len(df.columns)-1)
ylim = (0, df.drop("合计").values.max()*1.1)

years = list(df.columns)
x_ticks = np.arange(len(years))

# 只画竖直方向的网格线
for x in x_ticks:
    ax.axvline(x, color='#b0c4de', linewidth=1.2, zorder=1)

for idx, project in enumerate(df.index):
    if project != "合计":
        ax.plot(
            x_ticks,
            df.loc[project],
            marker='o',
            label=project,
            linestyle=linestyles[idx % len(linestyles)],
            color=colors[idx % len(colors)],
            zorder=2
        )


ax.set_xticks(x_ticks)
ax.set_xticklabels(years)
ax.set_xlim(xlim)
ax.set_ylim(ylim)
ax.set_title("国际科技合作项目趋势")
ax.set_xlabel("年份")
ax.set_ylabel("项目数量")
ax.legend(title="项目类型")
plt.tight_layout()
plt.show()
```
