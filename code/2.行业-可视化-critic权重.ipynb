1. 可及性：人工智能采纳程度

```python
import pandas as pd

# 读取采纳程度建模数据
df = pd.read_csv('/home/mw/input/data_total74037403/行业角度数据集/采纳程度建模数据.csv')
df_new = df[['年份', '采纳程度总和', '企业规模']]

# 读取企业数数据
df_inno = pd.read_csv('/home/mw/input/data_total74037403/行业角度数据集/全国企业创新调查-主表.csv')
df_inno_sub = df_inno[['年份', '企业规模', '企业数']]

# 合并企业数
df_merged1 = pd.merge(df_new, df_inno_sub, on=['年份', '企业规模'], how='left')

# 计算平均采纳程度
df_merged1['平均采纳程度'] = df_merged1['采纳程度总和'] / df_merged1['企业数']

print(df_merged1)
```

3.有效性：专利转化率

```python
import pandas as pd

# 读取专利建模数据
df = pd.read_csv('/home/mw/input/data_total74037403/行业角度数据集/专利转化率.csv')
df_new = df[['年份', '专利转化率', '企业规模']]

# 读取企业数数据
df_inno = pd.read_csv('/home/mw/input/data_total74037403/行业角度数据集/全国企业创新调查-主表.csv')
df_inno_sub = df_inno[['年份', '企业规模', '企业数']]

# 合并企业数
df_merged3 = pd.merge(df_new, df_inno_sub, on=['年份', '企业规模'], how='left')

# 只保留大型、中型、小型企业
df_merged3 = df_merged3[df_merged3['企业规模'].isin(['大型企业', '中型企业', '小型企业'])].reset_index(drop=True)

# 计算平均专利转化率
df_merged3['平均专利转化率'] = df_merged3['专利转化率'] / df_merged3['企业数']

print(df_merged3.head())
```

```python
# 合并三个DataFrame
df_all = pd.merge(df_merged1[['年份', '企业规模', '平均采纳程度']],
                  df_merged2[['年份', '企业规模', '平均专利']],
                  on=['年份', '企业规模'], how='outer')
df_all = pd.merge(df_all,
                  df_merged3[['年份', '企业规模', '平均专利转化率']],
                  on=['年份', '企业规模'], how='outer')

# 按照企业规模自定义顺序排序
scale_order = {'大型企业': 0, '中型企业': 1, '小型企业': 2}
df_all['企业规模排序'] = df_all['企业规模'].map(scale_order)
df_all = df_all.sort_values(['年份', '企业规模排序']).drop(columns='企业规模排序').reset_index(drop=True)

print(df_all.head())
```

```python
import numpy as np

# 只保留2016-2022年数据
df_all_sub = df_all[df_all['年份'].between(2016, 2022)].reset_index(drop=True)

# 只保留数值型指标部分
df_critic = df_all_sub[['平均采纳程度', '平均专利', '平均专利转化率']].copy()

# 标准化（极小化指标需反向，这里假设都是极大化指标）
df_norm = (df_critic - df_critic.min()) / (df_critic.max() - df_critic.min())

# 计算标准差
std = df_norm.std(axis=0)

# 计算相关系数矩阵
corr = df_norm.corr().values

# 计算CRITIC信息量
C = []
for j in range(len(std)):
    S = 0
    for k in range(len(std)):
        if j != k:
            S += (1 - corr[j, k])
    C.append(std[j] * S)
C = np.array(C)

# 归一化得到权重
weights = C / C.sum()
print("CRITIC权重：", weights)

# 计算科技平权指数
df_all_sub['科技平权得分'] = df_norm.values @ weights

print(df_all_sub[['年份', '企业规模', '科技平权得分']])
```

```python
import numpy as np

def gini_coefficient(x):
    x = np.sort(x)
    n = len(x)
    if n == 0:
        return np.nan
    cumx = np.cumsum(x)
    gini = (n + 1 - 2 * np.sum(cumx) / cumx[-1]) / n
    return gini

# 计算每年企业科技平权指数的基尼系数
gini_result = []
for year, group in df_all_sub.groupby('年份'):
    values = group['科技平权得分'].values
    gini = gini_coefficient(values)
    gini_result.append({'年份': year, '科技平权指数': gini})

df_gini = pd.DataFrame(gini_result)
print(df_gini)
```

```python
import matplotlib.pyplot as plt
import numpy as np

years = df_gini['年份']
values = df_gini['科技平权指数']

plt.figure(figsize=(10, 5))

# 用fill_between画折线到x轴之间的绿色渐变
n = 200
y_min = 0  # x轴
y_max = max(values)
for i in range(n):
    alpha = 0.55 * (1 - i / n)  
    color = (46/255, 139/255, 87/255, alpha)
    y_start = y_min + (y_max - y_min) * i / n
    y_end = y_min + (y_max - y_min) * (i + 1) / n
    plt.fill_between(years, y_start, np.minimum(values, y_end), 
                     where=(values >= y_start), color=color, zorder=0)

# 主曲线
plt.plot(years, values, marker='o', color='#4F77CE', linewidth=2, label='科技平权指数')

# 在折线旁边显示数据
for x, y in zip(years, values):
    plt.text(x, y, f'{y:.3f}', ha='center', va='bottom', fontsize=11, color='#4F77CE')

plt.xlabel('年份')
plt.ylabel('基尼系数')
#plt.title('不同类型企业科技平权指数（深绿色渐变面积图）')
plt.ylim(0.5, 0.6)
plt.xlim(2016, 2022)  
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
