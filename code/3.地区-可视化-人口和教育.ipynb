```python
import matplotlib
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# 导入数据
file_path = r'/home/mw/input/data_total74037403/地区角度数据集/人口和教育.csv'
data = pd.read_csv(file_path)

# 创建一个画布，包含四个子图
fig, axes = plt.subplots(2, 2, figsize=(20, 16))
features = ['2023高等学校R&D成员', '2023高等学校R&D课题经费（万元）', '2023高等学校发表科技论文', '2023高等学校有效发明专利']
titles = ['2023高等学校R&D成员分布', '2023高等学校R&D课题经费分布', '2023高等学校发表科技论文分布', '2023高等学校有效发明专利分布']

# 定义 data_sorted_by_patents 按照 '2023高等学校有效发明专利' 排序
data_sorted_by_patents = data.sort_values(by='2023高等学校有效发明专利', ascending=False)

# 遍历每个特征，绘制柱状图
for i, ax in enumerate(axes.flatten()):
    feature = features[i]
    sorted_data = data_sorted_by_patents.sort_values(by=feature, ascending=False)
    ax.bar(sorted_data['省份'], sorted_data[feature], color=plt.cm.tab20.colors)
    ax.set_title(titles[i], fontsize=16, fontweight='bold')
    ax.set_xticks(range(len(sorted_data['省份'])))
    ax.set_xticklabels(sorted_data['省份'], rotation=45, ha='right', fontsize=10)
    ax.set_ylabel(feature, fontsize=12)
    ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()

# 根据四个特征的排名计算综合排名
data_sorted_by_composite = data_sorted_by_patents.copy()
data_sorted_by_composite['综合排名'] = data_sorted_by_composite[['R&D成员排名', 'R&D课题经费排名', '发表科技论文排名', '有效发明专利排名']].mean(axis=1)
data_sorted_by_composite = data_sorted_by_composite.sort_values(by='综合排名')

# 创建一个热力图显示综合排名
fig, ax = plt.subplots(figsize=(12, 8))
heatmap_data = data_sorted_by_composite.pivot_table(index='省份', values='综合排名')
heatmap_data = heatmap_data.sort_values(by='综合排名', ascending=True)# 按排名排序
sns.heatmap(heatmap_data, cmap='YlGnBu', annot=False, cbar=True, ax=ax)

# 设置标题和轴标签
ax.set_title('综合排名热力图', fontsize=16, fontweight='bold')
ax.set_xlabel('')
ax.set_ylabel('省份', fontsize=12)

plt.tight_layout()

plt.show()
```

```python
# 按2023年（固定）互联网宽带接入用户（万户）排序
import pandas as pd
data = pd.read_csv('/home/mw/input/data_total74037403/地区角度数据集/地区.csv')
data_sorted_by_broadband = data.sort_values(by='2023年（固定）互联网宽带接入用户（万户）', ascending=False)

# 创建图表
fig, ax = plt.subplots(figsize=(18, 8))

# 绘制条形图
ax.bar(data_sorted_by_broadband['省份'], 
    data_sorted_by_broadband['2023年（固定）互联网宽带接入用户（万户）'], 
    color=plt.cm.Paired.colors, alpha=0.8, label='2023年互联网宽带接入用户（万户）')

# 设置x轴刻度标签
ax.set_xticks(range(len(data_sorted_by_broadband['省份'])))
ax.set_xticklabels(data_sorted_by_broadband['省份'], rotation=45, ha='right', fontsize=10)

# 添加数据标签
for i, value in enumerate(data_sorted_by_broadband['2023年（固定）互联网宽带接入用户（万户）']):
    ax.text(i, value, f'{value:.1f}', ha='center', va='bottom', fontsize=10, color='#4CAF50')

# 设置标题和轴标签
ax.set_title('各地区2023年互联网宽带接入用户分布（按宽带接入用户排序）', fontsize=16, fontweight='bold')
ax.set_xlabel('省份', fontsize=12)
ax.set_ylabel('互联网宽带接入用户（万户）', fontsize=12)

# 设置x轴刻度旋转
ax.set_xticks(range(len(data_sorted_by_broadband['省份'])))
ax.set_xticklabels(data_sorted_by_broadband['省份'], rotation=45, ha='right', fontsize=10)

# 添加网格和图例
ax.grid(axis='y', linestyle='--', alpha=0.7)
ax.legend(loc='upper right', fontsize=10)

# 美化布局
plt.tight_layout()

# 显示图表
plt.show()
```
