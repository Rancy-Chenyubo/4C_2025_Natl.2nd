```python
import pandas as pd

# 读取两个文件
df1 = pd.read_csv('/home/mw/input/data_total74037403/行业角度数据集/2023_stkcd_证券简称.csv', encoding='utf-8')
df2 = pd.read_csv('/home/mw/input/data_total74037403/行业角度数据集/2023_stkcd_股票简称.csv', encoding='utf-8')

# 统一列名
df1 = df1.rename(columns={'证券简称': '简称'})
df2 = df2.rename(columns={'股票简称': '简称'})

# 合并并去重
result = pd.concat([df1, df2], ignore_index=True)
result = result.drop_duplicates(subset=['stkcd'])

# 按照stkcd排序
result = result.sort_values(by='stkcd')

# 输出结果
print(result)
result.to_csv('2023_stkcd_合并.csv', index=False, encoding='utf-8')
```

```python
import matplotlib.pyplot as plt
from matplotlib import font_manager
# 设置支持中文的字体
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']  
plt.rcParams['axes.unicode_minus'] = False 
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# 读取数据
df = pd.read_csv('/home/mw/input/data_total74037403/行业角度数据集/人工智能采纳程度_合并后.csv')

# 只保留2016-2023年
df = df[(df['year'] >= 2016) & (df['year'] <= 2023)]

# 按年份和企业规模分组，计算人工智能采纳程度均值
grouped = df.groupby(['year', '企业规模'])['人工智能采纳程度'].mean().unstack()

# 绘图
fig, ax = plt.subplots(figsize=(8, 5))

# 获取坐标轴范围
xmin, xmax = df['year'].min(), df['year'].max()
ymin, ymax = grouped.min().min(), grouped.max().max()

# 生成渐变色（从上到下浅蓝色）
gradient = np.linspace(0, 1, 256).reshape(256, 1)
ax.imshow(
    np.hstack([gradient]*10), 
    extent=[xmin, xmax, ymin, ymax],
    origin='lower',
    aspect='auto',
    cmap=plt.get_cmap('Blues'),
    alpha=0.3,
    zorder=0
)

# 再画数据线
grouped.plot(marker='o', ax=ax, zorder=2,linewidth=2.5, markersize=6)


plt.grid(True, linestyle='--', alpha=0.7)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.title('2016-2022年各规模企业人工智能采纳程度均值')
plt.ylabel('人工智能采纳程度（均值）')
plt.xlabel('年份')
plt.legend(title='企业规模')
plt.tight_layout()
plt.show()
```
