```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import matplotlib as mpl


plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']  # 设置中文字体  
plt.rcParams['axes.unicode_minus'] = False

data = pd.read_csv('/home/mw/input/data_total74037403/地区角度数据集/经济与产业.csv')
provinces = data['Unnamed: 0']
gdp = data['地区生产总值（亿元）']
per_capita_gdp = data['人均地区生产总值（元）']
tertiary_value = data['第三产业增加值（亿元）']

# 按地区生产总值排序
sorted_data = data.sort_values(by='地区生产总值（亿元）', ascending=False)
provinces = sorted_data['Unnamed: 0']
gdp = sorted_data['地区生产总值（亿元）']
per_capita_gdp = sorted_data['人均地区生产总值（元）']
tertiary_value = sorted_data['第三产业增加值（亿元）']

# 图形设置
fig, ax1 = plt.subplots(figsize=(18, 10))

# 柱状图：地区生产总值
x = range(len(provinces))
bar_width = 0.4
bars = ax1.bar(x, gdp, bar_width, label='地区生产总值（亿元）', color='skyblue', edgecolor='black')
ax1.set_xlabel('省份', fontsize=12)
ax1.set_ylabel('地区生产总值（亿元）', fontsize=12)
ax1.set_title('各省地区生产总值、人均地区生产总值及第三产业增加值对比（按地区生产总值排序）', fontsize=16)
ax1.set_xticks(x)
ax1.set_xticklabels(provinces, rotation=45, ha='right', fontsize=10)

# 折线图：人均地区生产总值
ax2 = ax1.twinx()
line = ax2.plot(x, per_capita_gdp, label='人均地区生产总值（元）', color='orange', marker='o', linestyle='-', linewidth=2)
ax2.set_ylabel('人均地区生产总值（元）', fontsize=12)

# 气泡图：第三产业增加值
bubble_sizes = tertiary_value / 10
scatter = ax2.scatter(x, per_capita_gdp, s=bubble_sizes, label='第三产业增加值（亿元）', color='green', alpha=0.6, edgecolor='black')

# 图例设置
bars_legend = ax1.legend(loc='upper left', fontsize=10)
line_legend = ax2.legend(loc='upper right', fontsize=10)
scatter_legend = ax2.legend(['第三产业增加值（亿元）'], loc='lower right', fontsize=10)

# 网格线设置
ax1.grid(axis='y', linestyle='--', alpha=0.7)

# 显示图形
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import numpy as np
import matplotlib as mpl

import matplotlib.pyplot as plt

# 设置全局字体为微软雅黑
mpl.rcParams['font.sans-serif'] = ['Microsoft YaHei']
mpl.rcParams['axes.unicode_minus'] = False

# 数据准备
data = pd.read_csv('/home/mw/input/data_total74037403/地区角度数据集/经济与产业.csv', encoding='utf-8')

sorted_data = data.sort_values(by='地方一般公共预算支出', ascending=False)
provinces = sorted_data['Unnamed: 0']
budget_spending = sorted_data['地方一般公共预算支出']
education_ratio = (sorted_data['教育支出'] / sorted_data['地方一般公共预算支出']) * 100
science_ratio = (sorted_data['科学技术支出'] / sorted_data['地方一般公共预算支出']) * 100
authorization_count = sorted_data['授权数']
science_spending = sorted_data['科学技术支出']

# 图形设置
fig = plt.figure(figsize=(24, 12), constrained_layout=True)
gs = fig.add_gridspec(1, 2, width_ratios=[2, 1])

# 子图1：堆叠柱状图 - 一般公共预算支出及其科学技术和教育支出占比
ax1 = fig.add_subplot(gs[0, 0])
x = np.arange(len(provinces))
bar1 = ax1.bar(x, education_ratio, label='教育支出占比（%）', color='skyblue', edgecolor='black')
bar2 = ax1.bar(x, science_ratio, bottom=education_ratio, label='科学技术支出占比（%）', color='lightgreen', edgecolor='black')

ax1.set_xlabel('省份', fontsize=12)
ax1.set_ylabel('占比（%）', fontsize=12)
ax1.set_title('各省一般公共预算支出中教育和科学技术支出占比', fontsize=16)
ax1.set_xticks(x)
ax1.set_xticklabels(provinces, rotation=45, ha='right', fontsize=10)
ax1.legend(loc='upper right', fontsize=10)

# 添加数据标签
for i, (edu_ratio, sci_ratio) in enumerate(zip(education_ratio, science_ratio)):
    ax1.text(i, edu_ratio / 2, f'{edu_ratio:.1f}%', ha='center', va='center', fontsize=8, color='black')
    ax1.text(i, edu_ratio + sci_ratio / 2, f'{sci_ratio:.1f}%', ha='center', va='center', fontsize=8, color='black')

# 子图2：散点图 - 科学技术支出与专利授权数关系
ax2 = fig.add_subplot(gs[0, 1])
scatter = ax2.scatter(science_spending, authorization_count, s=authorization_count / 100, c=science_spending, cmap='coolwarm', alpha=0.8, edgecolor='black')
ax2.set_xlabel('科学技术支出（亿元）', fontsize=12)
ax2.set_ylabel('专利授权数', fontsize=12)
ax2.set_title('科学技术支出与专利授权数的关系', fontsize=16)

# 添加回归线
sns.regplot(x=science_spending, y=authorization_count, scatter=False, ax=ax2, color='red', line_kws={'linestyle': '--', 'linewidth': 2})

# 添加数据标签
for i, (x_val, y_val) in enumerate(zip(science_spending, authorization_count)):
    ax2.text(x_val, y_val, provinces.iloc[i], fontsize=8, ha='right', va='bottom')

# 调整颜色条
cbar = fig.colorbar(scatter, ax=ax2, label='科学技术支出（亿元）', orientation='vertical', pad=0.02)

# 显示图形
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib as mpl
import numpy as np

import matplotlib.pyplot as plt

# 设置全局字体为微软雅黑
mpl.rcParams['font.sans-serif'] = ['Microsoft YaHei']
mpl.rcParams['axes.unicode_minus'] = False

# 数据准备
data = pd.read_csv('/home/mw/input/data_total74037403/地区角度数据集/经济与产业.csv', encoding='utf-8')
data.rename(columns={
    '2023高技术产业企业数': 'High-Tech Enterprises',
    'Unnamed: 0': 'Province'
}, inplace=True)
sorted_data = data.sort_values(by='High-Tech Enterprises', ascending=False)
provinces = sorted_data['Province']
high_tech_enterprises = sorted_data['High-Tech Enterprises']

# 图形设置
fig, ax = plt.subplots(figsize=(18, 10))

# 绘制柱状图
x = range(len(provinces))
bars = ax.bar(x, high_tech_enterprises, color='skyblue', edgecolor='black')

# 设置标题和标签
ax.set_title('各省2023高技术产业企业数分布', fontsize=16)
ax.set_xlabel('省份', fontsize=12)
ax.set_ylabel('高技术产业企业数', fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels(provinces, rotation=45, ha='right', fontsize=10)

# 添加数据标签
for i, v in enumerate(high_tech_enterprises):
    ax.text(i, v + 50, str(v), ha='center', va='bottom', fontsize=8)

# 显示图形
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib as mpl

import matplotlib.pyplot as plt

# 设置全局字体为微软雅黑
mpl.rcParams['font.sans-serif'] = ['Microsoft YaHei']
mpl.rcParams['axes.unicode_minus'] = False

# 数据准备
data = pd.read_csv('/home/mw/input/data_total74037403/地区角度数据集/经济与产业.csv', encoding='utf-8')
data.rename(columns={
    'Unnamed: 0': 'Province',
    '规模以上工业企业R&D人员全时当量(人年)': 'R&D Personnel',
    '规模以上工业企业有效发明专利数（件）': 'Effective Patents',
    '规模以上工业企业R&D经费（万元）': 'R&D Funding'
}, inplace=True)
sorted_data = data.sort_values(by='R&D Personnel', ascending=False)
provinces = sorted_data['Province']
rd_personnel = sorted_data['R&D Personnel']
effective_patents = sorted_data['Effective Patents']
rd_funding = sorted_data['R&D Funding']

# 图形设置
fig, ax1 = plt.subplots(figsize=(18, 10))

# 主坐标轴：R&D人员全时当量和有效发明专利数
x = range(len(provinces))
line1 = ax1.plot(x, rd_personnel, label='R&D人员全时当量(人年)', color='blue', marker='o', linestyle='-', linewidth=2)
line2 = ax1.plot(x, effective_patents, label='有效发明专利数（件）', color='green', marker='s', linestyle='--', linewidth=2)
ax1.set_xlabel('省份', fontsize=12)
ax1.set_ylabel('数量', fontsize=12)
ax1.set_title('各省规模以上工业企业R&D人员全时当量、R&D经费与有效发明专利数的关系', fontsize=16)
ax1.set_xticks(x)
ax1.set_xticklabels(provinces, rotation=45, ha='right', fontsize=10)
ax1.legend(loc='upper left', fontsize=10)

# 次坐标轴：R&D经费
ax2 = ax1.twinx()
line3 = ax2.plot(x, rd_funding / 10000, label='R&D经费（千万元）', color='orange', marker='^', linestyle='-', linewidth=2)
ax2.set_ylabel('R&D经费（千万元）', fontsize=12)
ax2.legend(loc='upper right', fontsize=10)

# 添加数据标签
for i, (personnel, patents, funding) in enumerate(zip(rd_personnel, effective_patents, rd_funding)):
    ax1.text(i, personnel, f'{personnel}', ha='center', va='bottom', fontsize=8, color='blue')
    ax1.text(i, patents, f'{patents}', ha='center', va='bottom', fontsize=8, color='green')
    ax2.text(i, funding / 10000, f'{funding / 10000:.1f}', ha='center', va='bottom', fontsize=8, color='orange')

# 渐变背景
ax1.set_facecolor('#f7f7f7')
ax2.set_facecolor('#f7f7f7')

# 网格线设置
ax1.grid(axis='y', linestyle='--', alpha=0.7)

# 显示图形
plt.tight_layout()
plt.show()
```
