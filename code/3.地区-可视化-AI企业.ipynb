```python
import pandas as pd
from matplotlib import rcParams

import matplotlib.pyplot as plt

# 设置全局字体为微软雅黑
rcParams['font.sans-serif'] = ['Microsoft YaHei']
rcParams['axes.unicode_minus'] = False

# 数据导入
file_path = '/home/mw/input/data_total74037403/地区角度数据集/人工智能企业数量.csv'
data = pd.read_csv(file_path)

# 按AI企业数量排序
data_sorted = data.sort_values(by='AI企业数量总和（2000-2023）', ascending=False)

# 绘制柱状图
x = range(len(data_sorted))
bar_width = 0.5  # 调整柱间距离
plt.figure(figsize=(12, 8))
bars = plt.bar(x, data_sorted['AI企业数量总和（2000-2023）'], color='skyblue', width=bar_width)

# 添加数据标签，调整位置以避免遮挡
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, height + max(data_sorted['AI企业数量总和（2000-2023）']) * 0.01, 
             f'{height:,}', ha='center', va='bottom', fontsize=10, rotation=45)

# 设置标题和轴标签
plt.title('2000-2023年各省份AI企业数量总和', fontsize=16)
plt.xlabel('省份', fontsize=14)
plt.ylabel('AI企业数量', fontsize=14)

# 设置x轴刻度
plt.xticks(x, data_sorted['省份'], rotation=45, fontsize=10)

# 显示图表
plt.tight_layout()
plt.show()
```
