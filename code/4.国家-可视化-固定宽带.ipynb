```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib

# 设置全局字体为微软雅黑
matplotlib.rcParams['font.sans-serif'] = ['Microsoft YaHei']
matplotlib.rcParams['axes.unicode_minus'] = False

# 读取数据
df = pd.read_csv('/home/mw/input/data_total74037403/国家角度数据集/国家度量.csv')
df = df[df['国家名称'].notnull()]

plt.figure(figsize=(14, 7))
sns.violinplot(x='年份', y='固定宽带普及率', data=df, inner='box', scale='width', palette='YlGnBu')

# 计算每年均值并加标签
means = df.groupby('年份')['固定宽带普及率'].mean()
for i, year in enumerate(sorted(df['年份'].unique())):
    plt.text(i, means[year], f'{means[year]:.2f}', 
             ha='center', va='bottom', fontsize=12, color='black')

plt.title('2016-2022年全球主要国家固定宽带普及率分布（小提琴图）', fontsize=18)
plt.xlabel('年份', fontsize=15)
plt.ylabel('固定宽带普及率', fontsize=15)
plt.tight_layout()
plt.show()
```
