```python
import pandas as pd
import seaborn as sns

import matplotlib.pyplot as plt

# 读取数据
file_path = r'/home/mw/input/data_total74037403/地区角度数据集/地区.csv'
data = pd.read_csv(file_path)

# 设置全局字体为微软雅黑
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# 按国内专利授权数降序排序
data_sorted = data.sort_values(by='国内专利授权数', ascending=False)

# 创建图表
plt.figure(figsize=(12, 8))
sns.barplot(
    x='国内专利授权数',
    y='省份',
    data=data_sorted,
    palette='viridis'
)

# 添加数据标签
for index, value in enumerate(data_sorted['国内专利授权数']):
    plt.text(value + 500, index, f'{value}', va='center', fontsize=10)

# 设置标题和轴标签
plt.title('2023年各省份国内专利授权数分析', fontsize=16)
plt.xlabel('国内专利授权数', fontsize=12)
plt.ylabel('省份', fontsize=12)

# 美化图表
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()

# 显示图表
plt.show()
```
