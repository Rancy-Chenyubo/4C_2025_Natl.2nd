```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.patches import FancyBboxPatch
import numpy as np

# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False

# 数据
age_groups = ['50-59岁', '40-49岁', '30-39岁', '20-29岁', '6-19岁']
percentages = [10.7, 19.1, 23.9, 41.5, 21.1]
colors = ['#FF6347', '#FF1493', '#8A2BE2', '#4169E1', '#00CED1']

fig, ax = plt.subplots(figsize=(15, 8))

def draw_person(ax, x, y, color, alpha=1.0):
    """绘制简化的人形图标"""
    # 头部
    head = patches.Circle((x, y+0.1), 0.05, color=color, alpha=alpha)
    ax.add_patch(head)
    
    # 身体
    body_points = np.array([[x-0.03, y+0.05], [x+0.03, y+0.05], 
                           [x+0.04, y-0.1], [x-0.04, y-0.1]])
    body = patches.Polygon(body_points, color=color, alpha=alpha)
    ax.add_patch(body)
    
    # 手臂
    ax.plot([x-0.06, x-0.03], [y, y+0.02], color=color, linewidth=2, alpha=alpha)
    ax.plot([x+0.06, x+0.03], [y, y+0.02], color=color, linewidth=2, alpha=alpha)
    
    # 腿
    ax.plot([x-0.02, x-0.04], [y-0.1, y-0.2], color=color, linewidth=2, alpha=alpha)
    ax.plot([x+0.02, x+0.04], [y-0.1, y-0.2], color=color, linewidth=2, alpha=alpha)

max_icons = 20
for i, (age, pct, color) in enumerate(zip(age_groups, percentages, colors)):
    y_pos = i
    
    # 计算图标数量
    icon_count = int(pct / 50 * max_icons)
    
    # 绘制彩色图标
    for j in range(icon_count):
        x_pos = j * 0.3
        draw_person(ax, x_pos, y_pos, color, alpha=0.8)
    
    # 绘制灰色图标补充到满行
    for j in range(icon_count, max_icons):
        x_pos = j * 0.3
        draw_person(ax, x_pos, y_pos, '#CCCCCC', alpha=0.3)
    
    # 添加标签
    ax.text(-1, y_pos, age, fontsize=12, va='center', ha='right')
    ax.text(max_icons * 0.3 + 0.5, y_pos, f'{pct}%', fontsize=12, va='center')

ax.set_xlim(-2, max_icons * 0.3 + 2)
ax.set_ylim(-0.5, len(age_groups) - 0.5)
ax.axis('off')
ax.set_title('年龄分布人口图', fontsize=16, pad=20)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(10, 6))

# 数据
categories = ['男性', '女性']
values = [61.9, 56.2]
colors = ['#4F81BD', '#C0504D']  
icons = ['♂', '♀']

# 创建水平条形图，添加边框
bars = ax.barh(categories, values, color=colors, alpha=0.85, height=0.5, edgecolor='black', linewidth=1.5)

# 添加图标和数值
for i, (bar, icon, value, color) in enumerate(zip(bars, icons, values, colors)):
    ax.text(-7, i, icon, fontsize=40, ha='center', va='center', color=color, fontweight='bold')
    ax.text(value + 2, i, f'{value:.1f}%', fontsize=16, ha='left', va='center', color=color, fontweight='bold')


ax.set_xlim(0, 100)
ax.set_xlabel('百分比 (%)', fontsize=14)
ax.set_title('性别比例分布', fontsize=20, pad=15, fontweight='bold')
ax.grid(True, alpha=0.2, axis='x', linestyle='--')
ax.set_facecolor('#F7F7F7')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(1.2)
ax.spines['bottom'].set_linewidth(1.2)
ax.tick_params(axis='y', labelsize=14)
ax.tick_params(axis='x', labelsize=12)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

labels = ['高中及以下', '专科', '本科', '硕士及以上']
values = [43.7, 55.8, 61.6, 65.9]
colors = ['#fcd299', '#fdad5c', '#fd8e37', '#ff7031']

plt.figure(figsize=(9, 5))
radius = 1.0  # 所有水滴半径一致
for i, (label, value, color) in enumerate(zip(labels, values, colors)):
    center_x = i * 2.5
    center_y = 0

    # 画水滴外圈
    drop = plt.Circle((center_x, center_y), radius, color=color, alpha=0.25, linewidth=2, fill=True)
    plt.gca().add_patch(drop)
    # 画水滴边框
    border = plt.Circle((center_x, center_y), radius, color=color, fill=False, linewidth=2)
    plt.gca().add_patch(border)

    # 计算水位高度（-radius 到 +radius 线性映射 0~100%）
    water_height = -radius + (value / 100) * (2 * radius)

    # 画水位（波浪线填充）
    theta = np.linspace(np.pi, 0, 100)
    x_water = center_x + radius * np.cos(theta)
    # 波浪线
    wave_x = np.linspace(center_x - radius, center_x + radius, 100)
    wave_y = water_height + center_y + 0.06 * np.sin(6 * np.pi * (wave_x - (center_x - radius)) / (2 * radius))
    plt.fill_between(x_water, wave_y, center_y - radius, color=color, alpha=0.7)

    # 波浪水位线
    plt.plot(wave_x, wave_y, color=color, lw=2, alpha=0.9)

    # 数值标签
    plt.text(center_x, center_y, f"{value:.1f}%", ha='center', va='center', fontsize=15, color='#a14a00', fontweight='bold')
    # 类别标签
    plt.text(center_x, center_y - radius - 0.3, label, ha='center', va='top', fontsize=13)

plt.xlim(-1, (len(labels)-1)*2.5+1)
plt.ylim(-radius-1, radius+1)
plt.axis('off')
plt.title('不同学历层次占比（水滴水位图）', fontsize=16, pad=13)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ['一线城市', '新一线城市', '二三线城市', '四五线城市']
values = [58.1, 62.2, 58.5, 50.5]
colors = ['#bee7a5', '#94e38b', '#58d585', '#29bb74']

plt.figure(figsize=(7,4))
for i, (val, color) in enumerate(zip(values, colors)):
    plt.barh(i, val, color=color, height=0.68)
    plt.text(val+1, i, f'{val:.1f}%', va='center', fontsize=13, color='#19a148')
plt.yticks(range(len(labels)), labels)
plt.xlim(0, 75)
plt.xlabel('占比 (%)')
plt.title('不同城市等级占比', fontsize=16)
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
